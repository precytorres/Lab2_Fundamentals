```python
# Appendix 1 - Sensor Data Validator
# SEED: 4 Torres

SEED_NUM = 4

sensor_readings = [SEED_NUM - 2, SEED_NUM, SEED_NUM * 2, SEED_NUM * 5, SEED_NUM * 10]
# readings = [2, 4, 8, 20, 40]

def validate_reading(value):
    if value < 0:
        return "Critical Low"
    elif value <= 5:
        return "Low"
    elif value <= 20:
        return "Normal"
    elif value <= 35:
        return "High"
    else:
        return "Critical High"

print(f"SEED: {SEED_NUM}")
for reading in sensor_readings:
    status = validate_reading(reading)
    print(f"Reading {reading}: {status}")
```
